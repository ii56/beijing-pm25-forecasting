The table above presents the test-set performance of all models evaluated in this study.

### Summary of results:
- **Random Forest** achieves the best overall performance, with the lowest MAE (3.76), lowest RMSE (7.78), and highest R² (0.993).
- **Gradient Boosting** performs strongly and generalizes well, but remains slightly inferior to Random Forest.
- **Linear and Ridge Regression** provide solid baseline results, confirming that the engineered features capture much of the signal.
- **LSTM (PyTorch)** captures overall trends but produces significantly higher prediction error, indicating limited benefit from explicit sequence modeling for this dataset.

### Final model selection:
Random Forest is selected as the final model due to its superior accuracy, robust generalization, and lower complexity compared to deep learning approaches.
